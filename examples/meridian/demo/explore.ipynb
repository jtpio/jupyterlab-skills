# Solar noon over a year

The Sun does not cross the meridian at 12:00. The **equation of time** moves solar noon by up to *16 minutes* during the year.

- `equation_of_time(day)`: the shift, in minutes
- `solar_noon(day, longitude)`: the local clock time

```python
from datetime import date

from solartime import equation_of_time, load_places, solar_noon

places = load_places()
paris = places["Paris"]
print(f"{len(places)} places, Paris at {paris.longitude:.2f}°E")
```

```python
for month in range(1, 13, 2):
    day = date(2026, month, 1)
    noon = solar_noon(day, paris.longitude, paris.utc_offset)
    shift = equation_of_time(day)
    print(f"{day:%b %d}  noon {noon:%H:%M:%S}  shift {shift:+6.2f} min")
```
