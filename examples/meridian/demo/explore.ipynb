# Explore the stations

Load the stations and compare their monthly temperatures.

```python
from weather import load_stations, summary

stations = load_stations("stations.json")
[s.name for s in stations]
```

```python
for station in stations:
    print(f"{station.name:<8}", summary(station.temperatures))
```

## Warmest month

The month with the highest mean over all stations.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
means = [sum(s.temperatures[i] for s in stations) / len(stations) for i in range(12)]
months[means.index(max(means))], round(max(means), 1)
```
